```python
def lookup_to_geojson(lookup_dict):
    features = []

    for code, geometry in lookup_dict.items():
        feature = {
            "type": "Feature",
            "geometry": geometry,
            "properties": {
                "ISO_A3": code
            }
        }
        features.append(feature)

    return {
        "type": "FeatureCollection",
        "features": features
    }
```
