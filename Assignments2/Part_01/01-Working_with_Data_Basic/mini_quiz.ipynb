### 3️⃣ Quiz Question — Changing Tuple Elements

Given the following tuple:

```python
T = (13, 17, 31, 37, 73)
```

Write the necessary Python code that would change the second element of the tuple to 11.

````python
# Answer 3: tuples are immutable so we cannot change the second element of the tuple directly.
T_list = list(T) 
T_list[1] = 11  # Index 1 corresponds to the second element 
T = tuple(T_list) 
print(T) 
````

### 4️⃣ Quiz Question — Dictionary keys

Given the following dicitonary:

```python
D = {
    "apple":"10",
    "banana":"11",
    "cherry":"8"
}
```

Write the necessary Python code that would print the keys of the the dictionary D.

````python
# Answer 4
D = { 

    "apple":"10", 

    "banana":"11", 

    "cherry":"8" 

} 
for key in D:
    print(key)
````
