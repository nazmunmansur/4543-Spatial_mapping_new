```python

import pandas as pd

# Load the messy data
df = pd.read_csv("survey_dirty.csv")
df.head()
```

```python
df.info()
```

```python
# Strip whitespace from `name`, `email`, and other string columns
df['first_name'] = df['first_name'].str.strip()
df['last_name'] = df['last_name'].str.strip()
df['email'] = df['email'].str.strip()
```

```python
#Identify blank strings and convert them to NaN if needed
import numpy as np
df = df.replace(r'^\s*$', np.nan, regex=True)
print(df.isna().sum())
```

## 3. Handle missing values
- Identify columns with missing data
- Decide how to handle them: drop, fill, or flag (justify your choices)

```python
#Identify columns with missing data
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
df.fillna({
    "first_name": "Unknown",
    "last_name": "Unknown",
    "email": "Unknown",
    "exercise_freq": "Unknown",
    "height_cm": 0,
    "weight_kg ": 0,
    "satisfaction": "Unknown",
    "consent_given": "Unknown",
    "occupation": "Unknown",
    "city": "Unknown",
})
```

```python
df.dropna()# dropping out the missing values or values in the columns which has a value = nan
```

# 4. Standardize categorical data
- Convert `gender` values to consistent categories (`Male`, `Female`, `Other`)
- Normalize `exercise_freq` to a standard set of categories

```python
# Replace variations with standardized values
df['gender'] = df['gender'].replace(['M', 'm', 'male', 'Male'], 'Male')
df['gender'] = df['gender'].replace(['F', 'f', 'female', 'Female'], 'Female')
df['gender'] = df['gender'].replace(['Trans', 'Unknown', 'Other'], 'Other')

print(df)
```

```python
# Normalize `exercise_freq` to a standard set of categories
df['exercise_freq_update'] = df['exercise_freq'].str.lower().str.strip()
mapping = {'Low': 'S', 'moderate' : 'M', 'Moderate' : 'L','High' : 'XL',  'High' : '2XL',  'High' : '3XL'}
#df['exercise_freq'] = df['exercise_freq'].replace(mapping)
df['exercise_freq_update'] = df['exercise_freq'].map(mapping).fillna('Unknown')
print(df)
```
