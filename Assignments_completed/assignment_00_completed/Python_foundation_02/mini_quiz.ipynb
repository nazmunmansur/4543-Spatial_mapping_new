### 2️⃣ Given the following snippet:

```python
x = 7
y = 15
# z = ???
```
Write a boolean statement that would print true if the value in z was between x and y.

````python
# Answer 2
x = 7 
y = 15 
z = 12 

lies_between = True 
if x < z < y and lies_between: 
    print("True. z lies between the range") 
else: 
    print(False) 
````
